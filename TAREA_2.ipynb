```python
'''
NOMBRE: TORRES SANCHEZ ANGEL EDUARDO


Dada una lista de cadenas donde cada cadena representa una oracion
escirbir una funcion:
def indice_invertido(oraciones):
  .....
que retorne u diccionario donde:
1-las claves sean palabras
2-los valores sean conjuntos con los indices de las oraciones donde aparece la palabra
'''




def indice_invertido(oraciones):
  indice = {} # Diccionario vacío para guardar resultados
  for posicion , oracion in enumerate(oraciones):
    palabras = oracion.split() # Divide la oración en palabras
    for palabra in palabras:
      if palabra not in indice:
        indice[palabra] = []  # Crea una lista vacía para esa palabra
      indice[palabra].append(posicion)
  return indice  #devuelve el valor como diccionario

oraciones = ['python es un lenguaje', 'java es un lenguaje', 'python y java son populares']

print(indice_invertido(oraciones))



```

```python
'''
NOMBRE: TORRES SANCHEZ ANGEL EDUARDO

se tiene un diccionario:
cursos = {'matematicas':{'ana','luis','pedro'}
'fisica':{'ana','marta'}
'programacion':{'luis','pedro','marta'}}

escribir funciones para:
1-obtener el conjunto de estudiantes que estan inscritos en todos los cursos
2-obtener el conjunto de estudiantes que estan inscritos en al menos uno de los cursos
3-determinar si existen dos cursos con exctamente los mismos estudiantes
4-invertir el diccionario para obtener: {'ana':{'matematicas','fisica'}}
'''

cursos = {'matematicas':{'ana','luis','pedro'},
'fisica':{'ana','marta'},
'programacion':{'luis','pedro','marta'}}

#1-obtener el conjunto de estudiantes que estan inscritos en todos los cursos
inscritos_en_todos = set.intersection(*cursos.values())

print(inscritos_en_todos)

#2-obtener el conjunto de estudiantes que estan inscritos en al menos uno de los cursos
estudiantes = []
for x in cursos.values(): #metemos los valores en una lista
  estudiantes.append(x)

set1 = set()
for i in estudiantes: #metemos los valores en un set para hacer la operacion que le corresponde
  set1.update(i)

inscritos = set1 | set1 #conjunto de estudinates

print('todos los estudiantes ',inscritos)


#3-determinar si existen dos cursos con exctamente los mismos estudiantes
claves = list(cursos.keys())  # ← Esta línea es necesaria

for i in range(len(claves)):
    for j in range(i + 1, len(claves)):
        if cursos[claves[i]] == cursos[claves[j]]:
            print(f"{claves[i]} y {claves[j]} tienen los mismos estudiantes")
            break
    else:
        continue
    break
else:
    print("No hay cursos con los mismos estudiantes")

#4-invertir el diccionario para obtener: {'ana':{'matematicas','fisica'}}
invertido = {}
for curso, estudiantes in cursos.items():
    for estudiante in estudiantes:
        if estudiante not in invertido:
            invertido[estudiante] = set()
        invertido[estudiante].add(curso)

print(invertido)


```

```python
'''
NOMBRE: TORRES SANCHEZ ANGEL EDUARDO

Escribir una funcion:
def promedio_ponderado(calificaiones,pesos):

donde:
1-calificaciones es una lista de numeros
2-pesos es una lista del mismo tamaño
la funcion debe retornar el promedio ponderado
REQUISITOS:
1-si las listas tienen diferentes tama;o, debe lanzar valuerror
2-si la suma de los pesos no es 1 (con tolerancia pequeña), tambien debe lanzar valuerror
3-no modificar las listas originales
EJEMPLO:
promedio_ponderado([90,80,70],[0.5,0.3,0.2])
#--->83.0
'''

# Definir la función
def promedio_ponderado(calificaciones, pesos):
    if len(calificaciones) != len(pesos):
        raise ValueError("Las listas deben tener el mismo tamaño")

    suma_pesos = sum(pesos)
    tolerancia = 1e-6
    if abs(suma_pesos - 1.0) > tolerancia:
        raise ValueError("La suma de los pesos debe ser 1 (con tolerancia pequeña)")

    return sum(c * p for c, p in zip(calificaciones, pesos))

# Usarla
calificaciones = [90, 80, 70]
pesos = [0.5, 0.3, 0.2]

resultado = promedio_ponderado(calificaciones, pesos)
print(resultado)  # → 83.0
```

```python
'''
NOMBRE: TORRES SANCHEZ ANGEL EDUARDO

def aplicar_operacion(lista, funcion):
...
Que:
• Reciba una lista de números.
• Reciba una función que toma un número y retorna otro número.
• Devuelva una nueva lista con la función aplicada a cada elemento.
Luego definir tres funciones adicionales:
def cuadrado(x): ...
def cubo(x): ...
def absoluto(x): ...
Y demostrar su uso con aplicar_operacion.
'''

def aplicar_operacion(lista, funcion):
    return [funcion(x) for x in lista]

def cuadrado(x):
    return x ** 2

def cubo(x):
    return x ** 3

def absoluto(x):
    return abs(x)


numeros = [1, -2, 3, -4]
print(aplicar_operacion(numeros, cuadrado))   # [1, 4, 9, 16]
print(aplicar_operacion(numeros, cubo))       # [1, -8, 27, -64]
print(aplicar_operacion(numeros, absoluto))   # [1, 2, 3, 4]
```

```python
'''
NOMBRE: TORRES SANCHEZ ANGEL EDUARDO

Definir una función recursiva:
def suma_digitos(n):
...
Que reciba un entero no negativo y retorne la suma de sus dígitos.
Ejemplo:
suma_digitos(538)  # → 16
Restricciones:
• No convertir el número a cadena.
• No usar ciclos.
• Debe tener caso base claramente definido.
Luego explicar:
• ¿Cuál es el caso base?
• ¿Cuál es el caso recursivo?
'''

def suma_digitos(n):
    # Caso base: si el número es 0, la suma de sus dígitos es 0
    if n == 0:
        return 0
    # Caso recursivo: último dígito (n % 10) + suma de los demás dígitos
    return n % 10 + suma_digitos(n // 10)

# Ejemplo
print(suma_digitos(538))  # → 16
```
